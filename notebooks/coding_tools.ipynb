```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
SUBSCRIPTION = "Please state if you have a working subscription for an agentic coding tool?"
TOOL = "For which agentic coding tool do you have a subscrition?"
SAIA_KEY = "Do you need an GWDG SAIA API key?"
```

```python
df = pd.read_csv("coding_tools.csv")
# answer options in the export carry stray whitespace (" Other")
df[TOOL] = df[TOOL].str.strip()
```

```python
df.head()
```

```python
df[SUBSCRIPTION].value_counts(dropna=False)
```

```python
df[TOOL].value_counts(dropna=False)
```

```python
# participants without a subscription who still named a tool
pd.crosstab(df[SUBSCRIPTION], df[TOOL].fillna("no answer"))
```

```python
def pie(ax, counts, title):
    ax.pie(counts, labels=counts.index, autopct=lambda p: f"{p * counts.sum() / 100:.0f}",
           startangle=90, counterclock=False, wedgeprops=dict(edgecolor="white"))
    ax.set_title(f"{title}\n(n = {counts.sum()})")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
pie(axes[0], df[SUBSCRIPTION].value_counts(), "Working subscription for an\nagentic coding tool?")
# tools of everyone who named one, including 3 participants who answered "No" to a subscription
pie(axes[1], df[TOOL].value_counts(), "Agentic coding tool used")
pie(axes[2], df[SAIA_KEY].value_counts(), "Needs a GWDG SAIA API key?")
fig.tight_layout()

fig.savefig("coding_tools.pdf", bbox_inches="tight")
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 5))
pie(axes[0], df[SUBSCRIPTION].value_counts(), "Working subscription for an\nagentic coding tool?")
# tools of everyone who named one, including 3 participants who answered "No" to a subscription
pie(axes[1], df[TOOL].value_counts(), "Agentic coding tool used")
fig.tight_layout()

fig.savefig("coding_tools.png", bbox_inches="tight")
plt.show()
```
